```python
# Real-world Data Project (Retail Sales Analysis)

import pandas as pd
import matplotlib.pyplot as plt

# Sample sales data
data = {
    "Product": ["Laptop", "Mobile", "Tablet", "Headphones"],
    "Sales": [50000, 35000, 20000, 10000]
}

# Create DataFrame
df = pd.DataFrame(data)

# Print dataset
print(df)

# Total sales
total_sales = df["Sales"].sum()
print("\nTotal Sales:", total_sales)

# Bar chart
plt.bar(df["Product"], df["Sales"])
plt.title("Product Sales Analysis")
plt.xlabel("Products")
plt.ylabel("Sales Amount")
plt.show()
```
