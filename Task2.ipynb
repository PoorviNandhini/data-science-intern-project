```python
# Import libraries
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dataset
data = {
    'Hours': [1, 2, 3, 4, 5],
    'Marks': [20, 40, 50, 70, 90]
}

df = pd.DataFrame(data)

# Input and Output
X = df[['Hours']]
y = df['Marks']

# Create and train model
model = LinearRegression()
model.fit(X, y)

# Prediction
prediction = model.predict([[5]])

print("Predicted Marks:", prediction)
```
