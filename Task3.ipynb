```python
# Exploratory Data Analysis (EDA) - Small Program

import pandas as pd
import matplotlib.pyplot as plt

# Sample dataset
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [85, 90, 78, 92, 88]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display dataset
print("Dataset:\n")
print(df)

# Statistical Summary
print("\nStatistical Summary:\n")
print(df["Marks"].describe())

# Visualization
plt.bar(df["Student"], df["Marks"])
plt.title("Student Marks Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```
