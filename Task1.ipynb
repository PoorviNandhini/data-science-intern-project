```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample raw dataset
data = {
    "Name": ["Arun", "Bala", "Cathy", "David", "Eva", "Arun"],
    "Age": [25, 30, None, 45, 200, 25],
    "Salary": [25000, 30000, 28000, None, 1000000, 25000],
    "Department": ["HR", "IT", "IT", "Finance", None, "HR"]
}

# Create DataFrame
df = pd.DataFrame(data)

# Remove duplicates
df = df.drop_duplicates()

# Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["Department"] = df["Department"].fillna("Unknown")

# Remove outliers
df = df[df["Age"] <= 100]

# Display cleaned data
print("Cleaned Dataset:")
print(df)

# Graph 1: Bar Chart
plt.bar(df["Name"], df["Salary"])
plt.title("Salary by Employee")
plt.xlabel("Employee Name")
plt.ylabel("Salary")
plt.show()

# Graph 2: Pie Chart
dept_count = df["Department"].value_counts()
plt.pie(dept_count, labels=dept_count.index, autopct="%1.1f%%")
plt.title("Department Distribution")
plt.show()

# Graph 3: Line Chart
plt.plot(df["Name"], df["Age"], marker="o")
plt.title("Age by Employee")
plt.xlabel("Employee Name")
plt.ylabel("Age")
plt.show()
```
